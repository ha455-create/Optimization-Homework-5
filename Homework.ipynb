Covering Problems : 

An ambulance organization of a city operates 2 ambulances. These ambulances can be
stationed at any one of the 3 bases in the city. There are 5 districts in the city that needs
coverage. The table below shows which bases provide coverage to which districts, where an
entry of 1 corresponding to base i and district j indicates that an ambulance stationed at
base i can cover district j. For example, if we station an ambulance at base 2, then we
can cover districts 1, 3 and 5 with this ambulance. The populations of the 5 districts are
respectively 1500, 3500, 2500, 3000 and 2000. We assume that covering a district with more
than one ambulance does not bring any additional advantage over covering the district with
a single ambulance. In other words, having one ambulance stationed at a based that covers
a district is adequate to cover the population of the district. We want to decide where to
station the ambulances so that we maximize the total population under coverage.

```python
#importing all the modules 
import gurobipy as gp
from gurobipy import GRB
```

```python
# Problem data 
num_bases = 3
num_districs = 5
max_bases = 2
```

```python
# Coverage Matrix
coverage_matrix = [
    [1, 1, 0, 1, 0],  # Base 1
    [1, 0, 1, 0, 1],  # Base 2
    [0, 1, 0, 0, 1]   # Base 3
]
```

```python
#Population of each district
pop_matrix = [1500, 3500, 2500, 3000, 2000]
```

```python
#Optimization Model 
model = gp.Model("Ambulance_Coverage_Problem")
```

```python
# Decision Variables 
x = model.addVars(num_bases, vtype=GRB.BINARY, name="x")  # Binary variable for base selection
y = model.addVars(num_districs, vtype=GRB.BINARY, name="y")  # Binary variable for district coverage
```

```python
# Objective function to maximize total population covered 
model.setObjective(
    gp.quicksum(pop_matrix[j] * y[j] for j in range(num_districs)),
    GRB.MAXIMIZE
)
```

```python
# Coverage constraints in a for loop
for j in range(num_districs):
    model.addConstr(
        y[j] <= gp.quicksum(coverage_matrix[i][j] * x[i] for i in range(num_bases)),
        name=f"coverage_{j+1}"
    )
```

```python
# At most k bases can be selected
model.addConstr(
    gp.quicksum(x[i] for i in range(num_bases)) <= max_bases,
    name="base_limit"
)

```

```python
# Solving the model
model.optimize()

```

```python
# Printing optimal solution
if model.status == GRB.OPTIMAL:
    print("\nOptimal objective value : " , model.objVal)

    print("\nBases Selected:")
    for i in range(num_bases):
        print(f"Base {i+1}: {int(x[i].X)}")

    print("\nDistrict coverage:")
    for j in range(num_districs):
        print(f"District {j+1}: {int(y[j].X)}")
```
